Cada casilla tiene una figura y un valor asociado, que es dado al agente cuando visita la casilla. El agente debe guardar la información el valor acumulado recibido

```python
import random

class Environment:
    def __init__(self):
        self.board = [[random.randint(-100, 100) for y in range(0, 3)] for x in range(0, 3)]
        self.board[1][1] = 0
    
    
    def get_value(self, x, y) -> int:
        return self.board[x][y] 
      
```

## Prueba del agente

Ahora definimos la ejecución del agente, por defecto el agente ejecutará hasta que su valor acumulado sea mayor a 42. Una vez se alcance esta condición, imprimimos el valor total del agente y el número total de pasos a ejecutar. 

La ejecución del agente da un salto, actualiza el valor y luego retorna al agente a la posición original

```python
def main():
    episodes = 0
    env = Environment()
    a = Agent()
    print(env.board)
    tot_val = a.get_value()
    while tot_val < 42:
        action = a.choose_action()
        pos = a.do_action(action)
        val = env.get_value(pos[0], pos[1])
        a.add_value(val)
        tot_val = a.get_value()
        a.reset()
        episodes +=1
    print(f'El agente obtubo {tot_val} en {episodes} episodios')
        
main()
```
